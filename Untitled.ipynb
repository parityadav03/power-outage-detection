```python
import pandas as pd
import numpy as np
import joblib
from xgboost import XGBClassifier

# Load everything
scaler = joblib.load('scaler_hard.pkl')
xgb    = XGBClassifier()
xgb.load_model('xgb_model_native.json')

df     = pd.read_csv('features_day2.csv',
                      index_col='DateTime', parse_dates=True)
feature_cols = [c for c in df.columns if c != 'outage']
X    = df[feature_cols]
y    = df['outage']

# Rebuild hardened features
X['z1_z2_ratio']      = X['zone1'] / (X['zone2'] + 1e-6)
X['z1_z3_ratio']      = X['zone1'] / (X['zone3'] + 1e-6)
X['all_zones_sum']    = X['zone1'] + X['zone2'] + X['zone3']
X['zone_std']         = X[['zone1','zone2','zone3']].std(axis=1)
X['temp_x_hour']      = X['temp'] * X['hour']
X['wind_x_z1']        = X['wind_speed'] * X['zone1']
X['humidity_x_z1']    = X['humidity'] * X['zone1']
X['volatility_ratio'] = X['z1_rolling_std_w12'] / (X['z1_rolling_std_w144'] + 1e-6)
X['zscore_momentum']  = X['z1_zscore'].diff(6).fillna(0)

for col in ['z1_z2_ratio','z1_z3_ratio','all_zones_sum','zone_std',
            'temp_x_hour','wind_x_z1','humidity_x_z1',
            'volatility_ratio','zscore_momentum']:
    p1  = X[col].quantile(0.01)
    p99 = X[col].quantile(0.99)
    X[col] = X[col].clip(p1, p99)

X_scaled = scaler.transform(X)

# ── Test 1: Real outage rows from dataset ────────────────────
outage_idx = y[y == 1].index
print("="*50)
print("REAL OUTAGE ROWS — what model predicts on them")
print("="*50)
for i, idx in enumerate(outage_idx[:5]):
    row_pos  = df.index.get_loc(idx)
    row      = X_scaled[row_pos].reshape(1, -1)
    prob     = xgb.predict_proba(row)[0, 1]
    zscore   = float(X.loc[idx, 'z1_zscore'])
    zone1_raw= float(X.loc[idx, 'zone1'])
    print(f"  Row {i+1}: zone1={zone1_raw:.4f} | "
          f"z1_zscore={zscore:.4f} | prob={prob:.4f}")

# ── Test 2: Real normal rows ─────────────────────────────────
normal_idx = y[y == 0].index
print("\n" + "="*50)
print("REAL NORMAL ROWS — what model predicts on them")
print("="*50)
for i, idx in enumerate(normal_idx[:5]):
    row_pos = df.index.get_loc(idx)
    row     = X_scaled[row_pos].reshape(1, -1)
    prob    = xgb.predict_proba(row)[0, 1]
    zscore  = float(X.loc[idx, 'z1_zscore'])
    zone1_raw = float(X.loc[idx, 'zone1'])
    print(f"  Row {i+1}: zone1={zone1_raw:.4f} | "
          f"z1_zscore={zscore:.4f} | prob={prob:.4f}")

# ── Test 3: What does z1_zscore look like for outages? ───────
print("\n" + "="*50)
print("Z1_ZSCORE STATISTICS")
print("="*50)
print(f"  Normal rows mean zscore : {X.loc[y==0,'z1_zscore'].mean():.4f}")
print(f"  Normal rows std zscore  : {X.loc[y==0,'z1_zscore'].std():.4f}")
print(f"  Outage rows mean zscore : {X.loc[y==1,'z1_zscore'].mean():.4f}")
print(f"  Outage rows min zscore  : {X.loc[y==1,'z1_zscore'].min():.4f}")
print(f"  Outage rows max zscore  : {X.loc[y==1,'z1_zscore'].max():.4f}")
print(f"\n  zone1 feature stats in X:")
print(f"  zone1 mean : {X['zone1'].mean():.4f}")
print(f"  zone1 std  : {X['zone1'].std():.4f}")
print(f"  zone1 min  : {X['zone1'].min():.4f}")
print(f"  zone1 max  : {X['zone1'].max():.4f}")
```
